```python
from fastai.vision.all import *
import numpy as np
```

```python
import kagglehub
path = kagglehub.dataset_download("vipoooool/new-plant-diseases-dataset")

print("Path to dataset files:", path)
```

```python
dataset_path = Path("/kaggle/input/new-plant-diseases-dataset")

for p in dataset_path.rglob("train"):
    print("Train folder:", p)

for p in dataset_path.rglob("valid"):
    print("Validation folder:", p)
```

```python
train_path = Path("/kaggle/input/new-plant-diseases-dataset/new plant diseases dataset(augmented)/New Plant Diseases Dataset(Augmented)/train")
```

```python
dls = ImageDataLoaders.from_folder(train_path,valid_pct=0.2, seed=42, item_tfms=Resize(224))
```

```python
learner = vision_learner(dls, resnet18, metrics=error_rate)
learner.fine_tune(2)
```

```python
learner.export('/content/plant_disease_model.pkl')
```
